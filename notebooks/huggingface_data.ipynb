## Unify colums for each data frame

```python
# Get commons columns from both datasets in a set
common_columns = set(df_maharshipandya.columns).intersection(set(df_khepplewhite.columns))
common_columns
```

```python
common_columns.remove("Unnamed: 0")
common_columns
```

- Investigate track_id

```python
df_maharshipandya["track_id"]
```

```python
df_khepplewhite["uri"]
```

```python
df_khepplewhite["track_id"] = df_khepplewhite["uri"].str.split(":").str[-1]
df_khepplewhite.head()
```

- Artists

```python
def refactor_artists_namees(artists_names: str) -> str:
    split_names = artists_names.split(", ")

    
    # Unify with ;
    return ";".join(split_names)
```

```python
df_khepplewhite["artists"] = df_khepplewhite["artist_names"].apply(refactor_artists_namees)
df_khepplewhite.head()
```

```python
df_khepplewhite.head(-1)
```

```python
print(df_maharshipandya.columns)
print(df_khepplewhite.columns)
```

```python
common_columns.add("track_id")
common_columns.add("artists")
common_columns
```

```python
colums_for_maharshipandya = common_columns.copy()
colums_for_maharshipandya.add("duration_ms")
colums_for_maharshipandya.add("explicit")
colums_for_maharshipandya.add("track_genre")
colums_for_maharshipandya
```

```python
df_maharshipandya_filtered = df_maharshipandya[list(colums_for_maharshipandya)]
df_maharshipandya_filtered
```

```python
columns_for_khepplewhite = common_columns.copy()
columns_for_khepplewhite.add("language")
columns_for_khepplewhite
```

```python
df_khepplewhite_filtered = df_khepplewhite[list(columns_for_khepplewhite)]
df_khepplewhite_filtered
```

```python
# get index for each dataframe
df_maharshipandya_filtered.index
```

```python
df_khepplewhite_filtered.index
```

```python
# Combine both dataframe using track_id as key

df_maharshipandya_filtered = df_maharshipandya_filtered.set_index('track_id')
df_khepplewhite_filtered = df_khepplewhite_filtered.set_index('track_id')

```

```python
df_khepplewhite_filtered_unique_columns = list(set(df_khepplewhite_filtered.columns) - set(df_maharshipandya_filtered.columns))
df_khepplewhite_filtered_unique_columns
```

```python
df_merged = pd.merge(
    df_maharshipandya_filtered,
    df_khepplewhite_filtered[df_khepplewhite_filtered_unique_columns],
    left_index=True,
    right_index=True,
    how='outer'
)
```
